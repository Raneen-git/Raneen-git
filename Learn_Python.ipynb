%s - String (or any object with a string representation, like numbers)

%f - Floating point numbers
"%.number of digitsf" - Floating point numbers with fixed precision

Tuples
A tuple is a collection of different data types which is ordered and unchangeable (immutable). Tuples are written with round brackets, (). Once a tuple is created, we cannot change its values. We cannot use add, insert, remove methods in a tuple because it is not modifiable (mutable). Unlike list, tuple has few methods. Methods related to tuples:

tuple(): to create an empty tuple
count(): to count the number of a specified item in a tuple
index(): to find the index of a specified item in a tuple
+ operator: to join two or more tuples and to create a new tuple

```python
#age comparison
My_age = 25
your_age = int(input('Enter your age'))
if your_age > My_age:
  if your_age - My_age == 1:
    print("You are 1 year older than me")
  else :
      print("You are %d years older than me" %(your_age - My_age))
elif My_age > your_age:
  if My_age - your_age == 1:
    print("I am 1 year older than you")
  else :
      print("I am %d years older than you" %(My_age - your_age))
else :
  print("we are of the same age")



```

```python
#count number of evens and odds in an input of range of positive numbers
def count_even_odd(num):
  count_even = 0
  count_odd = 0
  for i in range(1, num +1):
    if i % 2 == 0:
      count_even += 1
    else:
      count_odd += 1
  return count_even, count_odd
print(count_even_odd(101))
```

```python
def factorial(number1):
  fact = 1
  for _ in range(1, number1 + 1):
    fact = fact * _
  return fact
print(factorial(5))
```

```python
#function which checks a number is prime or not
def is_prime(num):
  prime = True
  for i in range(2, num):
    if num % i == 0:
      prime = False
      break
    else:
      prime = True
  return prime
print(is_prime(9975))
```

```python
print(df_pharma.isna().sum())
#replace Na Values in Customer age as Mean age
mean_age = df_pharma['Customer_Age'].mean()
df_pharma.fillna({'Customer_Age' : mean_age}, inplace=True)
```

```python
#drop the missing date and unit price value
df_pharma.dropna(inplace=True)
print(df_pharma.isna().sum())

```

```python
#Formatting the dates
df_pharma['Date'] = pd.to_datetime(df_pharma['Date'], format= 'mixed')
print(df_pharma.to_string())
```

```python
#fixing duplicates
#finding duplicates
print(df_pharma.duplicated())
df_pharma.drop_duplicates(inplace=True)
print(df_pharma.duplicated())
```
